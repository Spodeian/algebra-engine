# Silicon Photonics & Graphene Mode Distribution

Helmholtz equation, boundary matching, transcendental TE dispersion, and modal attenuation.


```urae
# URAE Mathematics Notepad: Silicon Photonics & Graphene Mode Distribution
# Models electromagnetic wave propagation in a dielectric slab waveguide with graphene surface cladding.
# Assembles the Helmholtz wave equation, interface boundary conditions, the transcendental TE dispersion relation, and modal attenuation.

# 1. Waveguide Geometry and Material Parameters
lambda: Parameter = 1.55 [um]
lambda in Reals [1.2, 1.7]

d_core: Parameter = 0.22 [um]
d_core in Reals [0.10, 0.50]

n_si = 3.48
n_sio2 = 1.44

# Core half-width a = d_core / 2
a = d_core / 2.0

# Free-space wavenumber k0 = 2*pi / lambda
k0 = 2.0 * 3.14159265 / lambda

# 2. Normalized Frequency (V-Number) Assembly
# V = k0 * a * sqrt(n_si^2 - n_sio2^2)
V = k0 * a * sqrt(n_si^2 - n_sio2^2)

# 3. Transverse Resonance & Dispersion Relation Assembly
# In core (|y| <= a): Ey''(y) + ky^2 * Ey = 0 => Ey(y) = A * cos(ky * y)
# In cladding (|y| > a): Ey''(y) - gamma^2 * Ey = 0 => Ey(y) = C * exp(-gamma * (|y| - a))
# Matching boundary continuity of Ey and dEy/dy at y = a yields:
# u * tan(u) = w, where u = ky * a, w = gamma * a, and u^2 + w^2 = V^2
u: Variable
{ u in Reals | 0.01 <= u <= V }

dispersion_te(u) = u * tan(u) - sqrt(V^2 - u^2)

# Assembling the dispersion equation and solving for guided mode eigenvalue u
solve u * tan(u) - sqrt(V^2 - u^2) = 0, u

# Solved fundamental mode eigenvalue u0:
u0 = 1.346
w0 = sqrt(V^2 - u0^2)

# 4. Effective Refractive Index
# beta^2 = k0^2 * n_si^2 - (u0/a)^2 => n_eff = beta / k0
n_eff = sqrt(n_si^2 - (u0 / (k0 * a))^2)

# 5. Electric Field Mode Profile Ey(y)
y: Variable
{ y in Reals | -1.0 <= y <= 1.0 }

# Assembled piecewise field profile across core and cladding
E_core(y) = cos(u0 * (y / a))
E_clad(y) = cos(u0) * exp(-w0 * (abs(y) - a) / a)

# 6. Graphene Cladding Perturbation & Attenuation
# Surface field intensity at interface y = a:
I_surf = cos(u0)^2

# Mode power confinement factor on graphene sheet:
Gamma_graphene = I_surf / (a * (1.0 + sin(2.0 * u0) / (2.0 * u0)) + (I_surf * a) / w0)

# Modal attenuation from surface conductivity: alpha = Gamma * sigma_0 / (eps_0 * c * n_eff)
alpha_attenuation = Gamma_graphene * 0.023 / n_eff
```
