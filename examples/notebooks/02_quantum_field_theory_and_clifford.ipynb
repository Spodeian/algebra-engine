# URAE Tutorial: Quantum Field Theory & Space-Time Clifford Algebra

This tutorial demonstrates URAE's advanced high-energy physics engine:
- **Dirac Gamma Algebra**: $\{\gamma^\mu, \gamma^\nu\} = 2 \eta^{\mu\nu}\mathbf{I}_4$
- **Spinor Helicity Formalism**: Massless Weyl spinors $|p\rangle, |p]$ and Lorentz brackets $\langle i j \rangle, [i j]$
- **Parke-Taylor Tree Amplitudes**: $\mathcal{A}_n^{\mathrm{MHV}}(1^-, 2^-, 3^+, \dots, n^+) = \frac{\langle 1 2 \rangle^4}{\prod_{k=1}^n \langle k, k+1 \rangle}$
- **Passarino-Veltman 1-Loop Reductions**: $A_0(m^2), B_0(p^2, m_1^2, m_2^2), B_1(p^2, m_1^2, m_2^2)$

```python
import urae

# Evaluate traces of products of Dirac gamma matrices
# Tr[gamma^0 gamma^1 gamma^0 gamma^1] = -4
tr_4 = urae.qft.dirac_trace([0, 1, 0, 1])
print("Tr[γ^0 γ^1 γ^0 γ^1] =", tr_4)

# Traces of odd number of gamma matrices are identically zero
tr_3 = urae.qft.dirac_trace([0, 1, 2])
print("Tr[γ^0 γ^1 γ^2] =", tr_3)
```

```python
# Construct massless 4-momenta for 4 gluons in the center-of-mass frame
# p1 = (E, 0, 0, E), p2 = (E, 0, 0, -E), p3 = (E, E, 0, 0), p4 = (E, -E, 0, 0)
p1 = [100.0, 0.0, 0.0, 100.0]
p2 = [100.0, 0.0, 0.0, -100.0]
p3 = [100.0, 100.0, 0.0, 0.0]
p4 = [100.0, -100.0, 0.0, 0.0]

# Convert to Weyl 2-spinors
s1 = urae.qft.WeylSpinor.from_null_momentum(p1)
s2 = urae.qft.WeylSpinor.from_null_momentum(p2)
s3 = urae.qft.WeylSpinor.from_null_momentum(p3)
s4 = urae.qft.WeylSpinor.from_null_momentum(p4)

# Compute Mandelstam s_12 = <1 2> [2 1]
ang_12 = urae.qft.spinor_angle_bracket(s1, s2)
sq_21 = urae.qft.spinor_square_bracket(s2, s1)
s_12 = ang_12 * sq_21
print("Mandelstam invariant s_12 =", s_12)
```

```python
# Parke-Taylor 4-Gluon MHV Scattering Amplitude
amp = urae.qft.parke_taylor_4gluon_mhv([s1, s2, s3, s4], neg_helicity_indices=[0, 1])
print("A(1-, 2-, 3+, 4+) amplitude =", amp)
```
