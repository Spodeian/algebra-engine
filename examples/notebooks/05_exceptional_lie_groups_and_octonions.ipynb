# URAE Tutorial: Exceptional Lie Groups & Octonions

This tutorial explores exceptional algebraic structures in URAE:
- **Octonions $\mathbb{O}$**: 8D normed alternative division algebra with Fano plane multiplication
- **Associator $[x, y, z]$**: Non-associative measure $[x, y, z] = (xy)z - x(yz)$
- **Albert Exceptional Jordan Algebra $\mathcal{H}_3(\mathbb{O})$**: 27D Jordan algebra with Freudenthal cubic determinant
- **$E_8$ Root Lattice $\Gamma_8$**: 240 root vectors of length $\sqrt{2}$ and Weyl reflections $s_\alpha$

```python
import urae

# Construct three octonions
e1 = urae.exceptional.Octonion([0, 1, 0, 0, 0, 0, 0, 0])
e2 = urae.exceptional.Octonion([0, 0, 1, 0, 0, 0, 0, 0])
e4 = urae.exceptional.Octonion([0, 0, 0, 0, 1, 0, 0, 0])

# Compute product and associator [e1, e2, e4] = (e1 * e2) * e4 - e1 * (e2 * e4)
assoc = urae.exceptional.Octonion.associator(e1, e2, e4)
print("Associator [e1, e2, e4] =", assoc.coords)
print("Is non-zero?", assoc.norm() > 0.0)
```

```python
# Verify alternativity: [x, x, y] = 0 for any octonions x, y
alt_assoc = urae.exceptional.Octonion.associator(e1, e1, e2)
print("Alternator [e1, e1, e2] norm =", alt_assoc.norm())
```

```python
# Construct the E8 root lattice
roots = urae.exceptional.E8Lattice.all_roots()
print("Total E8 root vectors count =", len(roots))

# Verify that each root has squared norm = 2.0
sq_norms = [sum(c**2 for c in r) for r in roots[:5]]
print("First 5 root vector squared norms =", sq_norms)
```
