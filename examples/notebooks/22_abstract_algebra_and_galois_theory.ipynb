# URAE Tutorial: Abstract Algebra & Galois Theory

Examines field extensions and group symmetry:
- Solvability of quintic polynomials by radicals
- Galois group computation $\operatorname{Gal}(L/K)$
- Cyclotomic polynomials $\Phi_n(x)$ and character tables


```python
import urae

x = urae.var("x")
p = x**5 - 4*x + 2
grp = urae.galois_group(p, x)
print("Galois Group:", grp)
print("Solvable in radicals:", urae.is_solvable_by_radicals(p, x))
```
