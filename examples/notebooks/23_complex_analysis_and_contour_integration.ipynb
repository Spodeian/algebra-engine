# URAE Tutorial: Complex Analysis & Contour Integration

Complex function theory and Cauchy residue calculus:
- Laurent expansions $\sum_{n=-\infty}^\infty a_n (z - z_0)^n$
- Residue theorem $\oint_\gamma f(z)dz = 2\pi i \sum \operatorname{Res}(f, z_k)$
- Evaluation of stubborn real definite integrals


```python
import urae

z = urae.var("z")
f = urae.exp(z) / (z**2 * (z - 1))
print("Residue at z=0:", urae.residue(f, z, 0))
print("Residue at z=1:", urae.residue(f, z, 1))
```
