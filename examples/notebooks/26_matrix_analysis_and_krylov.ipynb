# URAE Tutorial: Matrix Analysis, SVD & Krylov Subspaces

Advanced matrix theory and sparse iterative solvers:
- Singular Value Decomposition (SVD) and Condition Numbers
- Schur triangular form and matrix exponential $e^A$
- Arnoldi iteration and GMRES linear equation solver


```python
import urae

A = urae.Matrix([[4.0, 1.0, 2.0], [1.0, 5.0, 3.0], [2.0, 3.0, 6.0]])
U, S, Vt = A.svd()
print("Singular values:", S)
print("Condition number:", A.cond())
```
