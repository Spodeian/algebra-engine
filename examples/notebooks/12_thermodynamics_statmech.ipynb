# URAE Tutorial: Thermodynamic Cycles & Statistical Mechanics

Welcome to the **Universal Rust Algebra Engine (URAE)** tutorial for thermodynamics, equations of state, Maxwell relations, and cyclic heat engine analyses.

$$\left(P + \frac{a}{V_m^2}\right)(V_m - b) = R T, \quad \eta_{\text{Carnot}} = 1 - \frac{T_C}{T_H}$$

This tutorial covers:
1. **Non-Ideal Gas Equations of State**: Van der Waals equation and critical state parameters ($T_c, P_c, V_c$)
2. **Maxwell Relations**: Analytical symbolic verification of $(∂S/∂V)_T = (∂P/∂T)_V$
3. **Carnot Engine Cycle**: Thermodynamic efficiency and P-V / T-S state diagram plots

```python
import urae

V = urae.var("V")
T = urae.var("T")
R_gas = 8.31446
a, b = 0.364, 4.27e-5 # Parameters for Nitrogen (N2)

# 1. Van der Waals Equation of State
P_vdw = (R_gas * T) / (V - b) - a / (V**2)
print("Van der Waals Pressure P(V, T) =", P_vdw)

# Critical point values
T_crit = (8 * a) / (27 * R_gas * b)
P_crit = a / (27 * (b**2))
V_crit = 3.0 * b
print(f"Critical State: T_c = {T_crit:.2f} K, P_c = {P_crit/1e6:.2f} MPa, V_c = {V_crit*1e3:.4f} L/mol")
```

```python
# 2. Analytical Maxwell Relation Verification: (∂P/∂T)_V = (∂S/∂V)_T
dP_dT = P_vdw.diff(T)
print("(∂P/∂T)_V =", dP_dT)
print("Consistent with (∂S/∂V)_T = R / (V - b):", dP_dT == (R_gas / (V - b)))
```

```python
# 3. Carnot Heat Engine Cycle Simulation
T_hot, T_cold = 650.0, 300.0
carnot_eta = 1.0 - (T_cold / T_hot)
print(f"Carnot Thermal Efficiency η = {carnot_eta * 100:.2f}%")

# Generate P-V and T-S Cycle Diagrams
urae.thermo.plot_carnot_pv(T_hot=T_hot, T_cold=T_cold, compression_ratio=4.0)
urae.thermo.plot_carnot_ts(T_hot=T_hot, T_cold=T_cold, entropy_change=25.0)
```
