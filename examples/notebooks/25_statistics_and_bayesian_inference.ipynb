# URAE Tutorial: Statistics, Probability & Bayesian Inference

Statistical computation and Bayesian inference:
- Probability density (PDF) and cumulative distribution (CDF)
- Conjugate Bayesian parameter updating
- Metropolis-Hastings Markov Chain Monte Carlo (MCMC)


```python
import urae

prior = urae.Beta(2.0, 2.0)
posterior = prior.update(successes=7, trials=10)
print("Posterior mean:", posterior.mean())
print("95% Credible Interval:", posterior.credible_interval(0.95))
```
