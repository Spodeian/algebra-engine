# Fluid Dynamics & Navier-Stokes Momentum Equations

Navier-Stokes laminar channel flow derivation and parabolic velocity profiles.


```urae
# URAE Mathematics Notepad: Fluid Dynamics & Incompressible Navier-Stokes
# Demonstrates assembling the Navier-Stokes momentum equations, integrating for laminar channel flow, matching boundary conditions, and computing shear stress.

# 1. Flow & Fluid Parameters
mu: Parameter = 0.001 [Pa*s]
rho: Parameter = 1000.0 [kg/m^3]
h: Parameter = 0.01 [m]
G: Parameter = 50.0 [Pa/m]

y: Variable
{ y in Reals | -h <= y <= h }

# 2. Assembling the 1D Steady Navier-Stokes Momentum Equation
# For steady laminar flow between parallel plates:
# mu * d^2 u / dy^2 = dP/dx = -G
# Integrating twice yields: u(y) = -(G / (2*mu)) * y^2 + C1 * y + C2
# Boundary conditions (no-slip): u(h) = 0, u(-h) = 0 => C1 = 0, C2 = G*h^2 / (2*mu)

# 3. Assembled Analytical Velocity Profile u(y)
u_channel(y) = (G * h^2 / (2.0 * mu)) * (1.0 - (y / h)^2)

# 4. Centerline Maximum Velocity & Mean Velocity
u_max = (G * h^2) / (2.0 * mu)
u_mean = (2.0 / 3.0) * u_max

# 5. Volume Flow Rate & Wall Shear Stress
# Volumetric flow rate per unit depth: Q = integrate(u(y), y, -h, h)
Q_flow = (2.0 * G * h^3) / (3.0 * mu)

# Wall shear stress tau_w = mu * |du/dy| at y = h
tau_wall = G * h

# 6. Reynolds Number Verification
Re_flow = (rho * u_mean * (2.0 * h)) / mu
```
