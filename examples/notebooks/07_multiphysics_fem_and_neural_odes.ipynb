# URAE Tutorial: Multiphysics Finite Elements & Neural ODEs

This tutorial explores continuum simulations and continuous deep learning in URAE:
- **2D/3D Finite Element Analysis (FEA)**: Poisson heat equation $-\nabla \cdot (k \nabla T) = Q$ and Navier-Cauchy linear elasticity
- **Symplectic Geometric Integrators**: 4th-order Yoshida energy-conserving phase space steppers
- **Neural Ordinary Differential Equations**: Continuous depth dynamical systems $\frac{d\mathbf{z}(t)}{dt} = f_\theta(\mathbf{z}(t), t)$ with Adjoint Sensitivity $\frac{d\mathbf{a}(t)}{dt} = -\mathbf{a}(t)^T \frac{\partial f}{\partial \mathbf{z}}$

```python
import urae

# Create structured 2D triangular FEA mesh on [0, 1] x [0, 1]
mesh = urae.pde.generate_uniform_grid_mesh(nx=10, ny=10, elem_type="Tri3")
print("FEA Mesh:", len(mesh.nodes), "nodes,", len(mesh.elements), "triangles.")
```

```python
# Solve 2D Steady-State Poisson Heat Conduction
# k = 45.0 W/(m*K), volumetric heat source Q = 1000 W/m^3
temperature = urae.pde.solve_poisson_heat(mesh, conductivity=45.0, heat_source=1000.0)
print("Peak Nodal Temperature T_max =", max(temperature), "K")
```

```python
# Run Symplectic 4th-order Yoshida integration on harmonic oscillator
traj = urae.quantum.SymplecticYoshida.integrate(
    q0=1.0,
    p0=0.0,
    dt=0.01,
    steps=1000
)
initial_energy = 0.5 * (1.0**2)
final_energy = 0.5 * (traj.q[-1]**2 + traj.p[-1]**2)
print("Energy Drift over 1000 steps =", abs(final_energy - initial_energy))
```
