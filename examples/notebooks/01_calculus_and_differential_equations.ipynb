# URAE Tutorial: Advanced Symbolic Calculus & Differential Equations

Welcome to the **Universal Rust Algebra Engine (URAE)** interactive notebook tutorial on calculus, limits, special functions, and ordinary differential equations (ODEs).

$$\mathcal{L}[y](t) = y''(t) + 2\beta y'(t) + \omega_0^2 y(t) = 0$$

```python
import urae

# Construct symbolic variables and parameters
x = urae.var("x")
t = urae.var("t")
a, b, omega = 2.5, 1.2, 3.14159

# Define an exponentially damped oscillator
f = a * urae.exp(-b * x) * urae.cos(omega * x)
print("Expression:", f)
```

```python
# Higher-order analytical derivatives
df_dx = f.diff(x)
d2f_dx2 = df_dx.diff(x)
print("df/dx =", df_dx)
print("d2f/dx2 =", d2f_dx2)
```

```python
# Analytical Risch Integration
poly_exp = (x**3) * urae.exp(x)
integral = poly_exp.integrate(x)
print("∫ x^3 e^x dx =", integral)
```

```python
# Multi-order L'Hôpital limits and Puiseux series
lim1 = urae.limit((1 - urae.cos(x)) / (x**2), x, 0)
ser1 = urae.series(urae.sin(x) / x, x, 0, order=6)
print("lim_{x->0} (1 - cos(x))/x^2 =", lim1)
print("Taylor series sin(x)/x =", ser1)
```
