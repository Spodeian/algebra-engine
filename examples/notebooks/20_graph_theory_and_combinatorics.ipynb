# URAE Tutorial: Spectral Graph Theory & Combinatorics

Explores discrete structures and algebraic graph theory:
- Adjacency $A$ and Degree $D$ matrices
- Graph Laplacian $L = D - A$ and algebraic connectivity $\lambda_2$
- Max-Flow Min-Cut and chromatic polynomial


```python
import urae

G = urae.Graph([(1, 2), (2, 3), (3, 4), (4, 1), (1, 3)])
L = G.laplacian()
conn = G.algebraic_connectivity()
print("Laplacian Matrix:\n", L)
print("Algebraic Connectivity lambda_2:", conn)
```
