# URAE Tutorial: Stochastic Itô Calculus & SDEs

Welcome to the **Universal Rust Algebra Engine (URAE)** tutorial on stochastic calculus, Wiener processes, quadratic variation, and Itô's Lemma.

$$df(t, X_t) = \left(\frac{\partial f}{\partial t} + \mu \frac{\partial f}{\partial x} + \frac{1}{2}\sigma^2 \frac{\partial^2 f}{\partial x^2}\right)dt + \sigma \frac{\partial f}{\partial x} dW_t$$

```python
import urae

# 1. Define drift and volatility parameters
mu = 0.05
sigma = 0.20
S = urae.var("S")
t = urae.var("t")

# 2. Formulate Geometric Brownian Motion (GBM)
# dS = mu * S * dt + sigma * S * dW
gbm = urae.stochastic.GeometricBrownianMotion(drift=mu, volatility=sigma, initial_value=100.0)
print("GBM Drift Term:", gbm.drift_expr(S))
print("GBM Diffusion Term:", gbm.diffusion_expr(S))
```

```python
# 3. Analytical Application of Itô's Lemma to log-price: f(S) = ln(S)
f_log = urae.ln(S)
df_dt_term, df_dw_term = urae.stochastic.ito_lemma(
    expr=f_log,
    var=S,
    drift=mu * S,
    diffusion=sigma * S
)
print("d(ln S) Drift component:", df_dt_term) # mu - 0.5 * sigma^2
print("d(ln S) Brownian component:", df_dw_term) # sigma
```

```python
# 4. Symbolic Black-Scholes PDE Operator
r = 0.03 # Risk-free rate
V = urae.function("V", [S, t])
bs_pde = V.diff(t) + r * S * V.diff(S) + 0.5 * (sigma**2) * (S**2) * V.diff(S).diff(S) - r * V
print("Black-Scholes PDE Operator:", bs_pde)
```
