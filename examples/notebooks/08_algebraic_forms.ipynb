# URAE Tutorial: Canonical Algebraic Forms & E-Graph Simplification

This notebook demonstrates algebraic transformations across different canonical forms:
- Expanded canonical monomial basis
- Factored irreducible forms
- Horner's nested polynomial representation for numerically stable evaluation
- Tropical semiring max-plus algebra

$$P(x) = a_0 + x(a_1 + x(a_2 + \dots + x a_n)), \quad x \oplus y = \max(x, y)$$

```python
import urae

x = urae.var("x")

# 1. Polynomial Expansion vs Factoring
p = (x + 2)**3 * (x - 5)**2
expanded_p = p.expand()
factored_p = expanded_p.factor()

print("Original:", p)
print("Expanded Form:", expanded_p)
print("Factored Form:", factored_p)
```

```python
# 2. Horner's Nested Rule for Optimal Flop Count & Numerical Stability
horner_expr = expanded_p.to_horner(x)
print("Horner Form:", horner_expr)
```

```python
# 3. Rational Simplification & Partial Fractions
rat = (2 * x + 1) / (x**2 - 1)
partial_frac = rat.apart(x)
recombined = partial_frac.together()
print("Rational:", rat)
print("Partial Fractions:", partial_frac)
print("Recombined:", recombined)
```

```python
# 4. Tropical Semiring Max-Plus Algebra
a, b = 12.0, 19.5
trop_sum = urae.tropical_add(a, b) # max(a, b)
trop_prod = urae.tropical_mul(a, b) # a + b
print(f"{a} ⊕ {b} =", trop_sum)
print(f"{a} ⊗ {b} =", trop_prod)
```
