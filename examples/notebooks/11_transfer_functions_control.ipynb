# URAE Tutorial: Transfer Functions & Control Engineering

Welcome to the **Universal Rust Algebra Engine (URAE)** tutorial for mechanical and electrical engineering control systems.

$$H(s) = \frac{Y(s)}{U(s)} = \frac{\omega_0^2}{s^2 + 2\zeta \omega_0 s + \omega_0^2}, \quad s_{1,2} = -\zeta \omega_0 \pm j \omega_0 \sqrt{1 - \zeta^2}$$

This notebook covers:
1. **Passive Electrical Networks**: Second-order RLC low-pass filter with Bode frequency response and pole-zero stability
2. **Active Electrical Networks**: Sallen-Key Op-Amp low-pass filter with adjustable gain
3. **Mechanical Engineering**: Damped mass-spring-damper vibration system

```python
import urae

s = urae.var("s")
R, L, C = 1000.0, 0.05, 1e-6

# 1. Passive RLC Second-Order Low-Pass Filter
H_rlc = 1.0 / (L * C * (s**2) + R * C * s + 1.0)
tf_rlc = urae.control.TransferFunction.from_expr(H_rlc, s)

print("RLC Transfer Function H(s) =", H_rlc)
print("System Poles:", tf_rlc.poles)
print("System Zeros:", tf_rlc.zeros)
print("Natural Frequency omega_0 =", tf_rlc.natural_frequency, "rad/s")
print("Damping Ratio zeta =", tf_rlc.damping_ratio)
print("Asymptotically Stable:", tf_rlc.is_stable)
```

```python
# Plot Bode Magnitude and Phase
urae.control.plot_bode(tf_rlc, omega_span=(10.0, 100000.0))
urae.control.plot_pole_zero(tf_rlc)
```

```python
# 2. Active Sallen-Key Op-Amp Low-Pass Filter
gain_K = 1.586 # Butterworth response for Sallen-Key
H_opamp = gain_K / (s**2 + (3.0 - gain_K) * s + 1.0)
tf_opamp = urae.control.TransferFunction.from_expr(H_opamp, s)

print("Op-Amp Transfer Function:", H_opamp)
print("Op-Amp Poles:", tf_opamp.poles)
urae.control.plot_step_response(tf_opamp, t_max=10.0)
```

```python
# 3. Mechanical Mass-Spring-Damper System
# m * x''(t) + c * x'(t) + k * x(t) = F(t)
m, k, c = 2.5, 250.0, 5.0
G_mech = 1.0 / (m * (s**2) + c * s + k)
tf_mech = urae.control.TransferFunction.from_expr(G_mech, s)

print("Mechanical Transfer Function G(s) =", G_mech)
print("Mechanical Poles:", tf_mech.poles)
print("Resonance Peak Frequency:", tf_mech.resonance_frequency, "rad/s")
```
