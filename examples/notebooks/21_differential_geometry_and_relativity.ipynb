# URAE Tutorial: Differential Geometry & General Relativity

Tensor calculus on curved Riemannian manifolds:
- Schwarzschild metric $g_{\mu\nu}$
- Christoffel connection coefficients $\Gamma^\lambda_{\mu\nu}$
- Riemann and Ricci curvature tensors


```python
import urae

coords = urae.symbols("t r theta phi")
M = urae.var("M")
metric = urae.schwarzschild_metric(coords, M)
gamma = metric.christoffel()
ricci = metric.ricci()
print("Ricci Scalar R:", metric.ricci_scalar())
```
